```python
#DAY-7 (GROUP BY)
import pandas as pd
emp_details = {
    "EmpID":[101,102,103,104,101],
    "City":["Hyderabad","Chennai","Bangalore","Vizag","Hyderbad"],
    "Experience":[3,4,3,5,2],
    "Age":[25,30,22,35,25]
}
df=pd.DataFrame(emp_details)
print(df)
```

```python
#Basic Groupby()
df.groupby('City')['Age'].mean()
```

```python
#Groupby on single column with multiple aggregations
df.groupby('City')['Age'].agg(['mean','max','min'])
```

```python
#Groupby On Multiple Columns With Multiple Aggregations
df.groupby('City').agg({
    'Age':['mean','max'],
    'Experience':['sum','count']
})
```

```python
# Merges and Joins (Left,right,inner,full outer)
df1=pd.DataFrame({
    'ID':[1,2,3],
    'Name':['A','B','c']
})
df2=pd.DataFrame({
    'ID': [1,2,3],
    'Salary':[1000,2000,3000]
})
print(df1)
print(df2)
```

```python
merged=pd.merge(df1,df2,on='ID',how='inner')
print(merged)
```

```python
merged=pd.merge(df1,df2,on='ID',how='outer')
print(merged)
```

```python
merged=pd.merge(df1,df2,on='ID',how='left')
print(merged)
```

```python
merged=pd.merge(df1,df2,on='ID',how='right')
print(merged)
```

```python
combined=pd.concat(
    [df1,df2],
    axis=0,
    ignore_index=True)
```

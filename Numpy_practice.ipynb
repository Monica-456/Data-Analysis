```python
#day 6 practice concatenation and stacking
#1D Concatenation:
#1.create 2 numpys and concate them into single 1d array
import numpy as np
a=np.array([10,20,30])
b=np.array([40,50,60])
result=np.concatenate((a,b))
print(result)
```

```python
#2.1D Concatenation-three arrays
#create a 1d arrays containing student marks for three sujects,concatenate all the three arrays into single array
import numpy as np
maths=np.array([98,97,96])
physics=np.array([87,86,85])
chemistry=np.array([76,78,79])
results=np.concatenate((maths,physics,chemistry))
print(results)
```

```python
#2D Horizontal
#3.given two arrays contain student_id and marks concatenate them horizontally using axis 1
import numpy as np
student_id=np.array([[101,102,103,104,105]]).T
marks=np.array([[98,87,78,89,76]]).T
result=np.concatenate((student_id,marks), axis=1)
print(result)
```

```python
#2D Vertical concatenation
#given two 2d arrays containing student records from two classes,concatenate them vertically using axis 0
import numpy as np
classA=np.array([101,102,103,104,105])
classB=np.array([106,107,108,109,110])
result=np.concatenate((classA,classB),axis=0)
print(result)
```

```python
#Horizontal concatenation-employee data
#create one array containing  employee names and another containig their salaries.combine them horizontally to create a single employee
#dataset
import numpy as np
names=np.array([["Satya Monica","Harika","Priya","Darshini","harsh"]]).T
salaries=np.array([[50000,60000,70000,80000,90000]]).T
result=np.concatenate((names,salaries),axis=1)
print(result)
```

```python
#Vertical Concatenation -sales Data    doubt
#create two arrays representing sales records for january and feburary.
#combine both vertically into one dataset
import numpy as np
january=([1000,2000,3000,4000,5000])
february=([2000,4000,6000,8000,9000])
total_sales=np.concatenate((january,february))
print(total_sales)
print(total_sales.ndim)
```

```python
#7. vertical stack practice
#create 3 1d arrays representing sales for three diff regions use np.vstack()
#to stack them into a 2d arrayb
mumbai=np.array([10,20,30])
chennai=np.array([40,50,60])
pune=np.array([70,80,90])
result=np.vstack((mumbai,chennai,pune))
print(result)
print(result.ndim)
```

```python
#8.hstack()
#create 2 arrays and combine them hrizontally
import numpy as np
employee_id=np.array([101,102,103,104])
salary=np.array([45000,55000,65000,75000])
result=np.hstack((employee_id,salary))
print(result)
```

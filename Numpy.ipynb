```python
#day 5
#aggregations 1D
#sum
import numpy as np
arr=np.array([10,20,30,40,50])
print(np.sum(arr))
print(np.min(arr))
print(np.max(arr))
print(np.mean(arr))
print(np.var(arr)) #variance
print(np.std(arr)) #standard deviation
print(np.percentile(arr,100)) #percentile=percentage=0 to 100
print(np.quantile(arr,0.50)) #quantile is always from 0 to 1.
print(np.argmax(arr)) #argmax returns the postion of the highest number unlike max which returns the number
print(np.argmin(arr)) #argmin returns the postion of the least number unlike min which returns the number
print(np.log10(arr))
```

```python
import numpy as np
arr=np.array([10,20,30,40,50,60]).reshape(2,3)
print(arr)
print(np.sum(arr))
print(np.min(arr))
print(np.max(arr))
print(np.sum(arr,axis=0))  #rows wise sum
print(np.sum(arr,axis=1))  #columns wise sum
print(np.mean(arr))
print(np.std(arr)) #standard deviation
print(np.var(arr)) #variance
print(np.percentile(arr,100))  #percentile =percentage=0 to 100
print(np.quantile(arr,1)) #quantile is always from 0 to 1.
print(np.argmax(arr)) #argmax returns the postion of the highest number unlike max which returns the number
print(np.argmin(arr)) #argmin returns the postion of the least number unlike min which returns the number, in 2d also shows the postion not like rows and columns(x,y)
```
